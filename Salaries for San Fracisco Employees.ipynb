```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("Total.csv")
df.head()
```

```python
print('There are {} Rows and {} Columns in the dataset.'.format(df.shape[0],df.shape[1]))
```

Data Cleaning

```python
df.info()
```

```python
series = ['BasePay','OvertimePay','OtherPay','Benefits']
for i in series:
    df[i] = pd.to_numeric(df[i], errors='coerce')

df.info()
```

Exploratory Data Analysis

```python
df['BasePay'].mean()
```

```python
df['BasePay'].describe()
```

```python
df.describe()
```

```python
df[df['BasePay']<0]
```

```python
job = df.groupby('JobTitle')['TotalPay'].mean().sort_values(ascending=False).head(10)
print(job)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=job.values, y=job.index, palette='Blues_d')
plt.title('Top 10 highest paying Job')
plt.xlabel('Average Total Pay')
plt.show()
```

Chief Investment Officer is the highest paying job having 500000 average total pay.

Distribution of BasePay vs OvertimePay vs OtherPay

```python
plt.figure(figsize=(20,5))

plt.subplot(1,3,1)
sns.histplot(df['BasePay'], bins=30, kde=True, color='blue')
plt.title('Distribution of BasePay')

plt.subplot(1,3,2)
sns.histplot(df['OvertimePay'], bins=30, kde=True, color='green')
plt.title('Distribution of OvertimePay')

plt.subplot(1,3,3)
sns.histplot(df['OtherPay'], bins=30, kde=True, color='red')
plt.title('Distribution of OtherPay')

plt.tight_layout()
plt.show()
```

Growth of Average Total Pay Over the Years

```python
pay = df.groupby('Year')['TotalPay'].mean()
plt.figure(figsize=(10,6))
sns.lineplot(x=pay.index, y=pay.values, markers='o', color='purple')
plt.title('Average Total Pay Over the Years')
plt.xlabel('Year')
plt.ylabel('Average Total Pay')
plt.show()
```

```python
plt.figure(figsize=(10,6))
corr_matrix = df[['BasePay', 'OvertimePay', 'OtherPay', 'Benefits', 'TotalPay', 'TotalPayBenefits']].corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Pay Components')
plt.show()
```
